**Q. 34. How do you encode and decode a string using Base64?**

Within `base64` package, we have `b64encode()` and `b64dencode()` methods

Example
```python
import base64

text = "India is my country"
encoded = base64.b64encode(text.encode())
decoded = base64.b64decode(encoded).decode()

print(encoded)
print(decoded)
```

**Q. 36. How do you remove duplicate characters from a string?**

```python
text = "How do you remove duplicate characters from a string?"
result = ""

for ch in text:
    if ch not in result:
        result += ch

print(result)
```

**Q. 37. How do you check if a string is a valid email address?**

```python
email = input("Enter email: ")

if (
    email.count("@") == 1
    and "@" in email
    and "." in email[email.index("@"):]
    and email[0] != "@"
    and email[-1] != "@"
    and email[email.index("@") + 1] != "."
    and email[-1] != "."
):
    print("Valid Email")
else:
    print("Invalid Email")
```
