# Python os Module - Questions and Answers

## Question 1
**What is the os module in Python?**

The **os** module provides functions to interact with the operating system. It is used to work with files, directories, environment variables, and system commands.

## Question 2
**What is the difference between mkdir() and makedirs()?**

| Function | Description |
|---|---|
| `os.mkdir()` | Creates a single directory. |
| `os.makedirs()` | Creates nested directories, including missing parent directories. |

## Question 3
**Explain remove() vs rmdir().**

| Function | Description |
|---|---|
| `os.remove()` | Deletes a file. |
| `os.rmdir()` | Deletes an empty directory. |

## Question 4
**What is the difference between rename() and replace()?**

| Function | Description |
|---|---|
| `os.rename()` | Renames a file or directory. |
| `os.replace()` | Replaces the destination if it already exists. |

## Question 5
**Why should you use os.path.join() instead of manually concatenating paths?**

`os.path.join()` creates platform-independent file paths using the correct path separator automatically.

## Question 6
**Explain os.walk() with an example.**

```python
import os
for root, dirs, files in os.walk('project'):
    print(root)
    print(files)
```
It recursively traverses all folders and files.

## Question 7
**What is the purpose of os.environ and os.getenv()?**

- `os.environ` stores environment variables as a dictionary.
- `os.getenv()` safely gets the value of an environment variable.

## Question 8
**What is the difference between os.listdir() and os.scandir()?**

| Function | Description |
|---|---|
| `os.listdir()` | Returns names of files and folders. |
| `os.scandir()` | Returns directory entries with additional information and better performance. |

## Question 9
**How do you check whether a file exists before deleting it?**

```python
import os
if os.path.exists('data.txt'):
    os.remove('data.txt')
```

## Question 10
**When would you use os.system()? What are its limitations?**

Use `os.system()` to execute simple OS commands.

**Limitations:**
- Limited error handling
- Returns only exit status
- Security risks with user input
- `subprocess` is preferred for most applications.
