## Question 30
**Create a menu-driven program demonstrating append, insert, remove, pop, sort, reverse, and display operations.**

```python
lst=[]
while True:
    print("\n1.Append 2.Insert 3.Remove 4.Pop 5.Sort 6.Reverse 7.Display 8.Exit")
    ch=int(input("Choice: "))
    if ch==1: lst.append(int(input("Value: ")))
    elif ch==2: lst.insert(int(input("Index: ")),int(input("Value: ")))
    elif ch==3: lst.remove(int(input("Value: ")))
    elif ch==4:
        if lst: lst.pop()
    elif ch==5: lst.sort()
    elif ch==6: lst.reverse()
    elif ch==7: print(lst)
    elif ch==8: break
```
