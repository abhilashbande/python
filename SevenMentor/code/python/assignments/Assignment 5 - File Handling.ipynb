# Assignment 5 – File Handling

### What is file handling in Python?

File handling in Python is the process of creating, opening, reading, writing, appending, and closing files. It allows programs to store and retrieve data permanently.

### What is the purpose of the open() function?

The open() function is used to open an existing file or create a new file. It returns a file object used for reading or writing.

Syntax:
file = open("filename.txt", "mode")

### What is the difference between r and r+?

- r : Opens a file for reading only. File must exist.
- r+ : Opens a file for both reading and writing. File must exist.

### What is the difference between w and a?

- w : Opens a file for writing and overwrites existing content
- a : Opens a file for appending without deleting existing content

### What is the difference between w and x?

- w : Creates a new file or overwrites an existing file
- x : Creates a new file only if it does not already exist

## Why should files be closed?

- Saves all changes.
- Frees system resources.
- Prevents data corruption.
- Avoids memory leaks.

### What is the advantage of the with statement?

The with statement automatically closes the file after use.

Example:
```python
with open("data.txt","r") as file:
    print(file.read())
```

### What is the difference between read(), readline(), and readlines()?

- read() : Reads the entire file
- readline() : Reads one line at a time
- readlines() : Reads all lines and returns a list

### What is the use of binary modes (rb, wb)?

- rb: Opens a file in binary read mode.
- wb: Opens a file in binary write mode.

Used for images, PDFs, audio, videos, etc.

### What exceptions can occur while opening files?

- FileNotFoundError
- PermissionError
- IsADirectoryError
- FileExistsError (when using x mode)
