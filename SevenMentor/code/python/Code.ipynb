### Variable name Rules
- Use only alphanumeric characters and underscores

- Case matters. The names age, Age, and AGE are treated as three completely distinct variables
- You cannot use Python's built-in keywords (such as if, else, while, for, class, or def) as variable names

## Type Casting

- Type casting is a process of converting a value from 1 data type to another
- This is done using type constructor such int(), float() and so on
- Python data type casting along with the type of casting
  1. Implicit
  2. explicit

### Implicit Casting
- It converts data types automatically if it is safe to do so

```python
a = 10
b = 40.34
print(a+b, type(a+b)) # int is getting converted into float automatically
```

### Explicit Casting
- It converts data types automatically if it is safe to do so

- It manually converts one object into another object using datatype constructor like int()

```python
str = '100'
iData = int(str) # explicit conversion to int
print(iData, type(iData))
```

### Arithmatic Operators

```python
productPrice = float(input("Enter Product Price = "))
print(productPrice)

shopDiscount = float(input("Enter shop discount (%) = "))
print(shopDiscount)

priceAfterDiscount = productPrice - productPrice * shopDiscount / 100;
print("Price after discount = ", priceAfterDiscount)

GST = float(input("Enter GST (%) = "));
finalProductValue = priceAfterDiscount + priceAfterDiscount * GST / 100;

print(f'Final product value after GST = {finalProductValue}')
```
