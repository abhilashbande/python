## Type Casting

- Type casting is a process of converting a value from 1 data type to another
- This is done using type constructor such int(), float() and so on
- Python data type casting along with the type of casting
  1. Implicit
  2. explicit

### Implicit Casting
- It converts data types automatically if it is safe to do so

````python
a = 10
b = 40.34
print(a+b, type(a+b)) # int is getting converted into float automatically
````

### Explicit Casting
- It converts data types automatically if it is safe to do so

- It manually converts one object into another object using datatype constructor like int()

````python
str = '100'
iData = int(str) # explicit conversion to int
print(iData, type(iData))
````

Question: Engineering and Medical Admission with Branch Selection (Nested if-else)

Write a Python program to check a student’s admission. First, accept the student’s overall percentage. If the percentage is 60% or above, ask the student to choose Engineering or Medical.
	•	If the student selects Engineering, ask them to choose one of the following branches:
	•	Computer Engineering (Minimum PCM: 85%)
	•	Mechanical Engineering (Minimum PCM: 75%)
	•	Civil Engineering (Minimum PCM: 70%)
	•	If the student selects Medical, ask them to choose one of the following branches:
	•	MBBS (Minimum PCB: 90%)
	•	BDS (Minimum PCB: 85%)
	•	BAMS (Minimum PCB: 75%)

````python
marks = float(input("\nEnter your score :: "))
isAllowed = False;
if marks >= 60:
    print("\nE. Engineering\nM. Medical")
    field = input("Enter field (E for Engineering or M for Medical) :: ").lower();

    if field == 'e' :
        print("\n1. Computer Engineering")
        print("2. Mechanical Engineering")
        print("3. Civil Engineering")
        branch = int(input("Enter branch code :: "))

        if branch == 1:
            if marks >= 85:
                isAllowed = True
        elif branch == 2:
            if marks >= 75:
                isAllowed = True
        elif branch ==3:
            if marks >= 70:
                isAllowed = True
        else:
            print("Error :: Invalid choice")
            
    elif field == 'm':
        print("\n1. MBBS")
        print("2. BDS")
        print("3. BAMS")
        branch = int(input("Enter branch code :: "))

        if branch == 1:
            if marks >= 90:
                isAllowed = True
        elif branch == 2:
            if marks >= 85:
                isAllowed = True
        elif branch ==3:
            if marks >= 75:
                isAllowed = True
        else:
            print("Error :: Invalid choice")
    else:
        print(f'Error :: {field} is not a valid choice')

if isAllowed:
    print("\nCongratulations!!! You are eligible for admission process\n")
else:
    print("\nOops... You can not join this field with this score\n")
````

### Jumping Control Statement

````python
balance = 10000

while True:
    print('''\n===== ATM MENU =====
    1. Check Balance
    2. Deposit Money
    3. Withdraw Money
    4. Exit''')

    choice = input("Enter your choice (1-4): ")

    if choice == "1":
        print(f"\nYour current balance is Rs. {balance}")

    elif choice == "2":
        amount = float(input("\nEnter amount to deposit: ₹"))

        if amount > 0:
            balance = balance + amount
            print(f"Rs. {amount} deposited successfully.")
            print(f"Updated balance: Rs. {balance}")
        else:
            print("\nInvalid deposit amount.")

    elif choice == "3":
        amount = float(input("\nEnter amount to withdraw: ₹"))

        if amount <= 0:
            print("\nInvalid withdrawal amount.")
        elif amount > balance:
            print("\nInsufficient balance.")
        else:
            balance = balance - amount
            print(f"\nRs. {amount} withdrawn successfully.")
            print(f"\nRemaining balance: Rs. {balance}")

    elif choice == "4":
        print("Thank you for using the ATM.")
        break

    else:
        print("Invalid choice. Please select between 1 and 4.")
````

## String Object

### Introduction

- A **string** is a sequence of characters enclosed in single (`' '`), double (`" "`), or triple quotes (`""" """`).
- Strings are **immutable**, meaning their characters cannot be changed after creation.
- Python supports **indexing** and **slicing** to access parts of a string.
- Common string methods include `upper()`, `lower()`, `strip()`, and `replace()`.

```python
name = "Python"
print(name[0])       # P
print(name.upper())  # PYTHON
```

#### Why string is an object???
- A String object contains data, method, properties


````python
str = "Strings are immutable, meaning their characters cannot be changed after creation"
print(f'Original String :: {str}')
````

### String Indexing

* **String indexing** is used to access individual characters in a string.
* Indexing starts from `0` for the first character.
* Negative indexing starts from `-1` for the last character.
* Use square brackets `[]` to access a character.


````python
text = "Python"
print(text[0])   # P
print(text[-1])  # n
````

### String Slicing

* **String slicing** is used to extract a portion of a string.
* It uses the `start`, `stop`, and optional `step` values.
* The `stop` index is **not included** in the result.
* Negative indexes can also be used for slicing.

#### Syntax
```python
string[start:stop:step]
```

````python
text = "India is my country."
print(text[0:3])   # Ind
print(text[::2])   # Idai ycuty
print(text[-2:-9:-1]) # yrtnuoc
````

### Immutable

* Strings in Python are **immutable**, meaning they cannot be changed after creation.
* You cannot modify an individual character using indexing.
* To make a change, Python creates a **new string** instead.

````python
text = "Python"
# text[0] = "J"  # TypeError

text = "J" + text[1:]
print(text)  # Jython
````

### String Methods

* **String methods** are built-in functions used to perform operations on strings.
* Common methods include `upper()`, `lower()`, `strip()`, `replace()`, and `split()`.
* Since strings are **immutable**, these methods return a new string instead of modifying the original string.

````python
text = "  Hello Python  "
print(text.upper())              # "  HELLO PYTHON  "
print(text.strip())              # "Hello Python"
print(text.replace("Python", "World"))  # "  Hello World  "
````
