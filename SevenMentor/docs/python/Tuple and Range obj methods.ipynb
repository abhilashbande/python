## 1. Tuple In Python :
    A Tuple is an ordered , Immutable (can not be changed after creation) collection of element .
    ----> Tuple is an Immutable version of list.

    key features :
        Ordered.
        Immutable.
        allow duplicate value
        can store diff diff type of obj in the tuple 
        Faster than list bcz it is an Immutable

### create tuple obj :
    Tuple is used when data should never changed

```python
# 1. empty tuple :
a=()
print(a,type(a),sep='\n')
```

```python
### with multiple elements 
days=('Monday','Tuesday','Wednesday','Thursday')
print(days,type(days),sep='\n')
```

### Tuple Packing 
    Packing means storing multiple values into the round bracket it means tuple

```python
student = 'Shiv','Mohan','Rohan','Rohini','kamal'
print(student,type(student),sep='\n')
```

## Tuple unpacking :
    unpacked each elemenmt into the each variable

```python
name1,name2,name3,name4,name5=student
print(name1)
print(name2)
print(name3)
print(name4)
print(name5)
```

## nested tuple 
    tuple inside the tuple 

```python
t=(('Shiv',28),('Rohan',45),('Mohan',33),('Komal',23))
print(t,type(t),sep='\n')
```

# tuple Methods:
    Is there only two methods:
            1. count
            2. index

### 1. Count :
    Return how many times a value occurs 

```python
a=(1,1,2,3,2,23,3,4,6,6)
a.count(1)
```

### 2. index :
    Return the first occurances index 

```python
print(a)
a.index(1)
```

# Range Data type :
    The Range data type represents an immutable seq of Number . It is mainly used with loops to generate numbers efficiently without storing all of them in memory.
    1. immutable
    2.memory efficient
    3. Generate number demanded
    4. Commonly used in for loops

    

```python
for i in range(11,20,1):
    print(i)
```
