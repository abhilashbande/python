# List :
    In python ,a list is a versatile and fundamental data structure used to store collections of items.
    Lists are mutable , ordered and allow elements of different data type.

#### key Featutres :
    - A list is a Built in data Structure in Python.
    - It is used to store collection of Items and entities.
    - Lists are enclosed in sq bracket [].
    - Elements in a list are seperated by Commas.
    - List can Store various data types , Including int, string ,float, even complex ......
    - List maintaine the order of elements . preserving the seq in which items were added.
    - Lists allow duplicates elements .
    - Lists are mutable, which means you can change their elements after the creation.

```python
# Creating list 
a=10   # int
b=10.23  # float
c='Shivsharan'  # str
d=12+45j   # complex
```

```python
listt=[a,b,c,d]
for i in listt:
    print(type(i),i)
```

# List slicing :
    List slicing allows to extract a subset of elememts from a list based on their indices.
    Syntax :
            variable[X:Y:Z]
        X--> start index
        y ---> end_index  (y-1)
        z ---> step size

```python
name=['Mohan','Rohan','shyam','Shiv','Anil']
name[0:4]=['Shivsharan']
```
