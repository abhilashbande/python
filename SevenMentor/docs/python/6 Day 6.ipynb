# Assignment Operator :
    This Operators are used to assign values to variables in Python .They allow you to store and Update data in variable .
        Here are the Commonly used assignment Operator In Python .
        1. = --> 
                Assignment --> Assign the value of The right operand to the left operand .
        2. +=  --> Add and Assign 
                Adds the right operand to the left operand and assigns the result to the left Operand .
        3. -= -->  Subtract and assign 
                subtract the right operand from the left Operand and assigns The result to the left operand .
        4. *= --> Multiply  and assignment 
        5. /= ---> Division and assign
        6. //= --> floor division and assign
        7. %=  --> Reminder and assign

```python
Number=100
print(f'Assign {Number} Value into the "Number" variable')
Number+=100
print(f'Add and Assign the Right operand into The left operand {Number}')
Number-=100
print(f'Subtract and assign the right operand from the left operand and value is = {Number}')
Number*=2
print(f'Multiply and assign {Number} ')
```

# 3. Logical operator :

        3.2 or :
                    This operator return true if atleast one of The operands is true, otherwise False.
        3.3 not :   invert the truth Value

# 5. Comparison Operator :
    It is called as conditional or relational operator .
    It Return  always boolean value{true/False}
    it is Binary Operator .

    ## Members of Comparison  Operator 
    (==,<=,>=,<,>,!= ,)
    == --->  Return True if the Both operand is equal
    != ---> Not equals to __ it check both operand not equals to each Other.
    < ----> Less than --> Check left Operand is Less than right Operand 
    > ----> More than ---> check left operand is More than right operand 
    <= ---> less than equals to --> check left operand is less and equal to right operand .
    >= ---> More than and equal to --> check left operand is more and equal to right operand .
    
    
    

```python
a=100
b=100.0
c=200
d=20
print(f'{a} and {b} is equals to each other  = {a==b}')
print(f'{a} and {c} is Not equal to each other = {a!=c}')
print(f'{d} is less than {a} = {d<a}')
print(f'{a} is more than {d} = {a>d}')
```
