#  Python Comments:
    Comment in python are Non-executed lines used to annotate and describe the code. They are ignored by the python Interpreter during runtime and are purely for Human readability .
    Main Purpose of comments are :
        1. Documentation
        2. Clarification
        3. Debugging

### Single Line Comments :
    Comments are denoted using the # 'Hash' symbol .
    Everything after the # on a line is considered a Comments and is ignored by the Python interpreter.

```python
 #### Single line comment 
# In this programm we use list object
a=[1,2,3,56,7,8,12,34,23,44,2,4,56,7]
# filter the list element if even then add 2 otherwise in this elemengt add 1
for i in a:
    if i%2==0 :  # in to the even number add 2 
        print(i+2)
    else:
        print(i+1)   # into the odd number add 1
```

### 2. Multiple Line Comments:
    MultiLine comments using the triple Quotes ('''Multiple lines''' or """Multiple Lines """ ). They often used for Multi-Line explanations or documentation strings{Docstring -->
    function_name.__doc__ } for function and classes.

```python
### Multiple line comment 
def even_odd(x):
    """
     ------ This programm for even and odd number manipulation ------
    put the list and then iterate the element on this 
    list and then filter the element .
    if element is even then  add in this element 2 otherwise 
    element is odd then add in this element 1 
    """
    for i in x:
        if i%2==0:
            'in this block even number filter'   # it is called as inline comment 
            print(i+2)
        else:
            'In this block odd number filter '
            print(i+1)
```

```python
print(even_odd.__doc__)
```

```python
even_odd()
```

```python
print()
```

### 3. In-Line Comment :
    This Comments are used to understanding instrction between the execte the code .

```python
x=[11,2,2,4,5,]
for i in x:
        if i%2==0:
            'in this block even number filter'   # it is called as inline comment 
            print(i+2)
        else:
            'In this block odd number filter '
            print(i+1)
```

# 2. Numerical Data type :
    2.1 int  :
        It is Base10 value without the decimal pointer from -ve infinity value to +ve infinity value such as 
        -10,0,10,100
        
    2.2 float :
        It is number with  decimal pointer value or in exponential from -ve infinity value to 
        +ve infinity value , such as 3.13 ,1.2,0.23,-0.124,-10,2e3
    2.3 complex  :
        Number with a real and imaginary part 
        written as a+bj , 
                where a and b are in the float form 
                and j is represent the imaginary unit 1j = sqrt(-1)

```python
a=-10
print(a,type(a),sep='\n')
```

```python
b=1
print(b,type(b),sep='\n')
```

```python
# float :
f=-0.29
print(f,type(f))
```

```python
f1=2e34
print(f1,type(f1))
```

```python
c=-10-20j
print(c,type(c))
```

```python
imag=c.imag
print(imag)
```

```python
real=c.real
print(real)
```
