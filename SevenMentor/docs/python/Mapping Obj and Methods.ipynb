# Dictonary :
    A dict is a Mutable built in Data Type that stores data as key and Value pairs.
    Each key is unique and maps to a value.

    Syntax :
        {keys:values}
    

```python
std_info={}
print(std_info,type(std_info))
```

```python
std_info={'Names' : ['Shiv','Mohan','Rohan','Anjana'],
         'Ages':[29,28,27,22],
         'Gender':('M','M','M','F')}
print(std_info)
```

```python
import pandas as pd
pd.DataFrame(std_info)
```

## Dict Methods :
    There are 11 commonly used dict method

```python
# 1. clear : Remove all items from dict
a={'a':10,'b':20}
print(f' a dict is {a}')
a.clear()
print(f'a dict is {a}')
```

```python
# 2. copy : create shallo copy new obj
a=std_info.copy()
print(a)
```
