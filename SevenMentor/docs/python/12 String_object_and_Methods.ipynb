### 6. strip() : 
    Return a copy of the string with leading and trailing whitespace removed.
    If chars is given and not None, remove characters in chars instead.

```python
name='___Shivsharan___'
print(name,len(name))
print('*'*55)
name.strip('_')  # remove characters in chars instead
```

```python
name='      Shivsharan Kunchalwar     '
print(name,len(name))
print('*'*55)
name.strip()   # bydefault whitespace remove from both side
```

### 7.lstrip and 8.rstrip :
    lstrip ---> Remove whitespaces from left side
    rstrip ---> Remove whitespaces from right side

```python
name='___Shivsharan___'
print(name,len(name))
print('*'*55)
name.lstrip('_')  # remove characters in chars instead
```

```python
name='      Shivsharan Kunchalwar     '
print(name,len(name))
print('*'*55)
name.rstrip()   # bydefault whitespace remove from right side
```
