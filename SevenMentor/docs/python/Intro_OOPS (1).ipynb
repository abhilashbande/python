```python
class EMS():
    def __init__(self,emp_id,name,salary,dept):
        self.emp_id=emp_id
        self.name=name
        self.salary=salary
        self.dept=dept
    def display(self):
        print(f'EMp id is {self.emp_id}')
    def calculate_salary(self,pf,tax):
        self.pf=pf
        self.tax=tax
        calculated_sal= self.salary -(pf+tax)
        print(f'emp calculated {self.salary -(pf+tax)}sal is {calculated_sal}')
    def updated_sal(self):
        print(f'emp sal is {self.salary -(self.pf+self.tax)}')
        
```

```python
TCS_EMS=EMS(1212,'Shivsharan',198000,'DS')
TCS_EMS.display()
```

```python
TCS_EMS.calculate_salary(1200,234)
```

```python
TCS_EMS.updated_sal()
```

# car class :
            ----------- Attributes ---------------
    brand 
    color
    model
    speed
    -----------------------
            ------------  Method ---------------------
    start()
    accerated()
    break()
    display_info()

```python
class car:
    def __init__(self, brand,color, model,speed):
        self.brand=brand
        self.color=color
        self.model=model
        self.speed=speed
    def start(self):
        print('car is starting')
    def accerate(self , increase):
        self.speed=self.speed+increase
        print('accerate = ',self.speed)
    def break_(self,decrease):
        self.speed=self.speed-decrease
        print('Break',self.speed)
    def display_info(self):
        print(f'Brand :{self.brand}')
        print(f'Color = {self.color}')
        print(f'model is {self.model}')
        print(f'speed is {self.speed}')
```

```python
Tata_Punch=car('Tata_punch','red',2026,350)
Tata_Punch.start()
Tata_Punch.accerate(0)
Tata_Punch.break_(0)
```
