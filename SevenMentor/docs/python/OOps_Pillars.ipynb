# Oops Pillars :
    The 4 Pillars of OOps are :
        1. encapsulation 
            -->  Wrapping data and methods together inside a class and controlling access to them .
        2. Inheritance
            --> Creating a new class from an existing class to reuse its properties and methods.
        3. Polymorphism :
            --> One interface/ method name can behave differently for different obj.
        4. Abstraction :
            --> hiding complex implementation details and showing only essentail functionality.

```python
class Vehicle() :
    def start(self):
        print('Vehicle starts')

class car(Vehicle):   # inheritance 
    def start(self):  # polymorphism 
        print('car starts')
        
```

```python
car=car()
car.start()
```

# Single  Inheritance :
    When one child class inherits from one another class it is called Single Inheritance .

```python
class User:   # parent class 
    def __init__(self,name,email):
        self.name=name
        self.email=email
    def login(self):
        print('User logged in')
    def logout(self):
        print('User Logged Out')
        
class Cutomer(User):  # Child class 
    def PlaceOrder(self):
        print('Order Placed')
    def CancelOrder(self):
        print('order Cancelled')
```

```python
obj_1=Cutomer('Shiv','gmail')
obj_1.CancelOrder()
obj_1.login()
obj_1.logout()
obj_1.PlaceOrder()
```

# 2. Multiple inheritance :
    When One child class inherits from multiple parent classes, It is called as Multiple Inheritance

```python
class Prgrammer:
    def write_code(self):
        print('Writing Python code')
class DataScentist:
    def analyze_data(self):
        print('Analyzing data')
class AIEngineer(Prgrammer,DataScentist):
    def Build_model(self):
        print('Building AI model')
```

```python
obj_model=AIEngineer()
obj_model.write_code()
obj_model.analyze_data()
obj_model.Build_model()
```

# 3. Multilevel Inheritance :
     When Inheritance happens ina chain of Multiple levels it is called Multilevel inheritance

```python
class person:
    def eat(self):
        print('Person Can eat')
class employee(person):
    def work(self):
        print('EMP is working')
class Manager(employee):
    def Manage_Team (self):
        print('Manager manage Team ')
```
