# File Opening modes :
    1. read mode ---> 'r'
    used to read an existing file 


```python
product_file=open(r'C:\Users\Admin\Documents\SQL\SQL_Text_data\Product.txt','r')
print(product_file.read())
product_file.close()
```

    2. write mode ----> 'w'
        create new file if it does not exists.
        if the file is already exists . It deletes all exisiting content before writing .

```python
New_file=open(r'student.txt','w')
New_file.write('''
im shiv im from Pune 
currently work as a DS trainer @7mentor 
......
....
...
''')
New_file.close()
```

```python
New_file=open(r'C:\Users\Admin\Documents\SQL\SQL_Text_data\student.txt','w')
New_file.write('''
im shiv im from Pune 
currently work as a DS trainer @7mentor 
......
....
...
''')
New_file.close()
```

    3. append () :
        Adds new data at the end of the file
        existing data remains unchanged .

```python
New_file=open(r'student.txt','r')
print(New_file.read())
```

```python
New_file=open(r'student.txt','a')
New_file.write('\n1234567werty')
New_file.close()
```

```python
New_file=open(r'student.txt','r')
print(New_file.read())
```

    4. read and write mode (r+)
        read and write both

```python
f=open('student.txt','r+')
print(f.read())
f.write('\n Data Science')
f.close()
```

```python
f=open('student.txt','r+')
print(f.read())
```

    5. write and read (w+)    
        Create a new file .
        Delete previous content
        

```python
f2=open('student.txt','w+')
f2.write('.....')
f2.seek(0)
print(f2.read())
f2.close()
```

    append and read ---> a+
    --> Adds data and also allows reading.

```python
f3=open('student.txt','a+')
f3.write('''
im shiv im from Pune 
currently work as a DS trainer @7mentor 
......
....
...''')
f3.seek(0)
print(f3.read())
f3.close()
```

### writing methods :

###### writelines  
    write a Multiple string 

```python
f=open('student.txt','w')
lines=['Shiv','\nMohan','\nRohan','\nSoham']
f.writelines(lines)
f.close()
```
