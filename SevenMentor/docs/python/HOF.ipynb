# Higher Order Function :
    A HOF is a function that takes another function as an argument, return a function as a result , or both .

```python
def add(a,b):  # Normal function 
    return a+b
```

```python
def calculate(func,a,b):  # Higher order function 
    return func(a,b)
```

```python
calculate(add,10,23)
```

# Built in  HigherOrder Function In python :
    map 
    filter 
    reduce
    

# map :
    map applies a function to every elements of an iterable.

    Syntax :
        map(function, iterable_obj)

```python
def square_(a):
    return a**2
```

```python
sq= lambda x: x**2 
list(map(sq,[1,2,3,4]))
```

```python
list(map(square_,[1,2,3]))
```

```python
num=[1,2,3,5]
mult=lambda x: x*x 
list(map(mult,num))
```

```python
add_list= lambda x,y : x+y
list(map(add_list,[1,2,3],[4,5,6]))
```

```python
even_number = lambda x: x if x%2==0 else x+1
```

```python
list(map(even_number,[1,2,3,10,5]))
```

# filter :
    Filter() select element based on the condition 

        filter(function , iterable)

```python
even_number_ = lambda x:  x%2==0 
list(filter(even_number_,[1,2,3,4]))
```

# reduce :
    reduce is avilable through the function 
    it work seq on iterable from obj
    

```python
def add(a,b):
    return a+b
```

```python
list(map(add,[1,2,3],[2,3,4]))
```

```python
from functools import reduce
```

```python
reduce(add,[1,2,3])
```

```python
reduce(lambda x,y: x*y,[1,2,3,4])
```
