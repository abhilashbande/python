    sep ---> 
        Sep stands for separator .
        It is used to define  what should be placed between multiple values inside the print() function , By default one space is placed .

    end ---> 
        It specifies what should be printed after the output is displayed.


    Syntax :
         print(*args, sep=' ', end='\n', file=None, flush=False)
