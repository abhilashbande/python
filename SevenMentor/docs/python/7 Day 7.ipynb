# 6.identity Operator :
    In Python , Identity operator are used to compare the Memory addresses of two obj to determine whether they refer to the same obj in Memory.

    There are two identity operators in  Python  :
        6.1 is 
        6.2 is not

### 6.1 is operator :
    The is Operator return True if Both operand refer to the same obj in Memory ,Otherwise it return False

```python
x='Shivsharan'
y='Shivsharan'
print(id(x),id(y), 'is same')
x is y
```

```python
a=100000
b=100000
print(id(a),id(b))
```

```python
a=10.0
b=10.0
id(a),id(b)
a is  b 
```

### 6.2 is not operator :
    The is not operator return True if both operands do not refer to the same obj in memory , Otherwise it return False

```python
a='Hello'
b='hello'
c='Elli'
print(a is not b)
print(a is not c)
```

# 7. Bitwise Operator :
    Bitwise Operator in Python are used to perform operation on Individual bits of integers.These operators are particularly useful when you need to manipulate binary data or perform Low Level Bitwise Operation.

    Here are the Bitwise operators avilable in python.
    Bitwise and ===>   &
    Bitwise or ===>    |
    Bitwise not ===>   ~
    left Shift ===>    <<
    Right Shift ===>   >>

#### 7.1 Bitwise and ===>   & :
    The Bitwise and Operator compare Two Numbers bit by bit and return 1 where both bits are 1.

```python
10 & 11
```

```python
print(bin(11))
print(bin(13))
print('-'*7)
print(bin(11&13))
print('_'*7)
print(11&13)
```

```python
bin(22)
```

```python
15 & 17
```

### 7.2 Bitwise OR :
    ----> | 
    perform a Bitwise or Operator on each pair of corresponding bits in two integers . 
    if at least one of the bits is 1 then return is 1.

```python
15 | 17
```

```python
print(bin(11))
print(bin(13))
print('-'*7)
print(bin(11|13))
print('_'*7)
print(11|13)
```

### 7.3 Bitwise XOR 
        ---> ^  {Caret} exclusive OR operation.
        Perform a bitwise exclusive OR (XOR) operation on each pairs of corresponding bits in two integers .
        If the bits are different the result is 1 otherwise 0.

```python
print(bin(11))
print(bin(13))
print('-'*7)
print(bin(11^13))
print('_'*7)
print(11^13)
```
